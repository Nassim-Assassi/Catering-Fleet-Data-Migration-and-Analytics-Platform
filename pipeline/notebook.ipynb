```python
import pandas as pd
from paths import TRANSFORMED_DATA_FILE
import matplotlib.pyplot as plt
```

```python
try:
    print(f"Reading transformed data from {TRANSFORMED_DATA_FILE}...")
    df = pd.read_csv(TRANSFORMED_DATA_FILE, parse_dates=['at'])
    print(f"Successfully read transformed data")
except Exception as e:
    print(f"Failed to read transformed data: {e}")
```

```python
df.dtypes
```

```python
# plot the distribution of average sales per hour of the day as a line graph

df['hour'] = df['at'].dt.hour

plt.figure(figsize=(10, 6))
plt.plot(df.groupby('hour')['total'].mean().sort_index())
plt.title('Distribution of Average Sales per Hour of the Day')
plt.xlabel('Hour of the Day')
plt.ylabel('Average Sales')
plt.xticks(range(24))  
plt.show()
```

```python
# plot the distribution of average number of sales per hour of the day as a line graph with one line for each truck name

plt.figure(figsize=(10, 6))

for truck_name, group in df.groupby('truck_name'):
    plt.plot(group.groupby('hour').size().sort_index(), label=truck_name)

plt.title('Distribution of Average Number of Sales per Hour of the Day by Truck')
plt.xlabel('Hour of the Day')
plt.ylabel('Average Number of Sales')
plt.xticks(range(24))
plt.legend()
plt.show()
```
